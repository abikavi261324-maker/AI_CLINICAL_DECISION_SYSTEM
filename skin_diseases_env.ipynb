```python
import cv2
import numpy as np

images = []
labels = []

for i in range(len(df)):
    img = cv2.imread(df['path'].iloc[i])
    img = cv2.resize(img, (224, 224))
    img = img / 255.0
    
    images.append(img)
    labels.append(df['dx'].iloc[i])

    if len(images) >= 500:   # keep limit for system
        break

images = np.array(images)
labels = np.array(labels)

print("Images:", images.shape)
print("Labels:", labels.shape)
print("Classes:", set(labels))
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
labels_encoded = le.fit_transform(labels)

print("Encoded labels:", set(labels_encoded))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    images, labels_encoded, test_size=0.2, random_state=42
)

print(X_train.shape, X_test.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

model = Sequential()

model.add(Conv2D(32, (3,3), activation='relu', input_shape=(224,224,3)))
model.add(MaxPooling2D(2,2))

model.add(Conv2D(64, (3,3), activation='relu'))
model.add(MaxPooling2D(2,2))

model.add(Flatten())

model.add(Dense(128, activation='relu'))
model.add(Dense(len(set(labels_encoded)), activation='softmax'))

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.summary()
```

```python
history = model.fit(
    X_train, y_train,
    epochs=5,
    validation_data=(X_test, y_test)
)
```

```python
images = np.array(images)
labels = np.array(labels)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
labels_encoded = le.fit_transform(labels)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    images, labels_encoded, test_size=0.2, random_state=42
)
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

datagen = ImageDataGenerator(
    rotation_range=20,
    zoom_range=0.2,
    horizontal_flip=True
)

datagen.fit(X_train)
```

```python
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D

base_model = MobileNetV2(
    weights='imagenet',
    include_top=False,
    input_shape=(224,224,3)
)

model = Sequential([
    base_model,
    GlobalAveragePooling2D(),
    Dense(128, activation='relu'),
    Dense(7, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    datagen.flow(X_train, y_train, batch_size=32),
    epochs=10,
    validation_data=(X_test, y_test)
)
```

```python
import numpy as np

test_img = X_test[0]
test_img = np.expand_dims(test_img, axis=0)

prediction = model.predict(test_img)
predicted_class = le.inverse_transform([np.argmax(prediction)])

print("Prediction:", predicted_class)
```

```python
if predicted_class[0] == 'mel':
    print("⚠️ Skin Cancer Risk! Please consult a doctor.")
else:
    print("Normal or less risky condition.")
```
